```python
import matplotlib.pyplot as plt
from scipy.stats import norm
import numpy as np
```

```python
import matplotlib.pylab as pylab
params = {'legend.fontsize': 'x-large',
          'figure.figsize': (15, 5),
         'axes.labelsize': 'x-large',
         'axes.titlesize':'x-large',
         'xtick.labelsize':'x-large',
         'ytick.labelsize':'x-large'}
pylab.rcParams.update(params)
```

```python
# generating multiple normal distributions
domain = np.linspace(0, 100, 1000) # dividing the distance between -2 and 2 into 1000 points

means = [33, 72, 50]
std_values = [10, 10, 10]

plt.figure(figsize=(16, 9))
for mu, std in zip(means, std_values):
    # pdf stands for Probability Density Function, which is the plot the probabilities of each range of values
    probabilities = norm.pdf(domain, mu, std)
    plt.plot(domain, probabilities, label=f"$\mu={mu}$\n$\sigma={std}$\n")

plt.legend()
plt.xlabel("Value",fontsize=16)
plt.ylabel("Probability",fontsize=16)
plt.savefig('std.png', dpi=100)
plt.show()
```

```python
bins = [0,20,40,60,80,100]
plt.hist(domain, bins, facecolor='r', alpha=0.7, edgecolor='k', linewidth=1)
plt.title("Primes between 1 and 100")
plt.xlabel("Primes")
plt.ylabel("Frequency")
plt.show()
```
