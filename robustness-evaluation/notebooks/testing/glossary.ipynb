```python
with open("test.txt","r", encoding="utf8") as f1: 
    data = f1.readlines()
```

```python
temp=[]
for x in data:
    y = x.split("=")
    s1 = y[1]
    s2 = y[0]
    temp.append("\gloss{"+s1+"}{"+s2+"}")
```

```python
new_list = [s.replace("\n", "") for s in temp]
print ("List in proper method", '[%s]' % ', '.join(map(str, new_list)))
```
